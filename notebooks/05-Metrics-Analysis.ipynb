```python
import pandas as pd
import numpy as np

import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.io as pio
# from fpdf import FPDF

import glob
import os
```

```python
# Get the list of all the files in the folder
files = glob.glob('outputs/04*.csv')

# Create a list of dataframes
dfs = [pd.read_csv(f) for f in files]

# Add the name of the file to the dataframe
for i, df in enumerate(dfs):
    df['file'] = os.path.basename(files[i])

# Concatenate all the dataframes
df = pd.concat(dfs)

# Sort the dataframe by the file name
df = df.sort_values('file')

# Reset the index
df.reset_index(drop=True, inplace=True)

df
```

```python
# Rename file column for type = 'univariate'
df.loc[df['type'] == 'univariate', 'file'] = '-'
```

```python
df_multivariate = df[df['type'] == 'multivariate']

# Sort the dataframe by scenario
df_multivariate = df_multivariate.sort_values('scenario')

# Reset the index
df_multivariate.reset_index(drop=True, inplace=True)

df_multivariate
```

```python
# Keep one row of the scenario and type == 'univariate'
df_univariate = df[df['type'] == 'univariate']

# Keep only one row that the values match except by the file name
df_univariate = df_univariate.drop_duplicates(subset=['scenario', 'R2', 'MAPE', 'MAE', 'RMSE'])

# Sort the dataframe by scenario
df_univariate = df_univariate.sort_values('scenario')

# Reset the index
df_univariate.reset_index(drop=True, inplace=True)

df_univariate
```

```python
# Rename the 'name' column from df_univariate to 'benchmark-univariate'
df_univariate['name'] = "benchmark"
```

```python
# Create df that is the concatenation of df_univariate and df_multivariate
df = pd.concat([df_univariate, df_multivariate])
```

```python
df.to_csv('outputs/all_stats.csv', index=False)
```

```python
# Remove the 'CCY', 'type', 'file' columns
df = df.drop(columns=['CCY', 'type', 'file'])

# Sort by scenario and name
df = df.sort_values(['scenario', 'name'])

# Reset the index
df.reset_index(drop=True, inplace=True)

df
```

```python
# # Create a new column for x-axis
# df['x_axis'] = df['name'] + '-' + df['scenario'].astype(str) + '-' + df['type']

# # Metrics to plot
# metrics = ['MAPE', 'MAE', 'RMSE']

# # Number of scenarios
# num_scenarios = df['scenario'].nunique()

# # Create subplots: 1 row for each scenario, 1 column for each metric
# fig = make_subplots(rows=num_scenarios, cols=len(metrics), subplot_titles=[f"{metric} Scenario {i+1}" for i in range(num_scenarios) for metric in metrics])

# # Add traces for each metric
# for i, scenario in enumerate(df['scenario'].unique()):
#     for j, metric in enumerate(metrics):
#         scenario_data = df[df['scenario'] == scenario].sort_values(by=metric)
        
#         # Get min and max values for colour scaling
#         min_value = scenario_data[metric].min()
#         max_value = scenario_data[metric].max()
        
#         # Create the colour scale
#         color_scale = px.colors.sequential.Plasma
        
#         # Map the metric values to colours
#         colors = px.colors.sample_colorscale(color_scale, (scenario_data[metric] - min_value) / (max_value - min_value))
        
#         # Add trace with the coloured bars
#         fig.add_trace(
#             go.Bar(x=scenario_data['x_axis'], y=scenario_data[metric], name=f"{metric} Scenario {scenario}", marker_color=colors),
#             row=i+1, col=j+1
#         )

# # Update layout
# fig.update_layout(height=800*num_scenarios, width=1200, title_text="Metric Plots by Scenario", showlegend=False)

# # Show plot
# fig.show()

# Create a new column for x-axis
df['x_axis'] = df['name'] + '-' + df['scenario'].astype(str) + '-' + df['type']

# Metrics to plot
metrics = ['MAPE', 'MAE', 'RMSE']

# Number of scenarios
num_scenarios = df['scenario'].nunique()

# Generate a unique colour for each x_axis element
unique_x_axis_elements = df['x_axis'].unique()

# Generate a list of unique colours
color_scale = px.colors.qualitative.Alphabet  # This provides a larger set of distinct colours
if len(unique_x_axis_elements) > len(color_scale):
    # If there are more elements than colours available, we need to generate more colours
    color_scale = color_scale * (len(unique_x_axis_elements) // len(color_scale) + 1)

color_mapping = {element: color_scale[i] for i, element in enumerate(unique_x_axis_elements)}

# Create subplots: 1 row for each scenario, 1 column for each metric
fig = make_subplots(rows=num_scenarios, cols=len(metrics), subplot_titles=[f"{metric} Scenario {i+1}" for i in range(num_scenarios) for metric in metrics])

# Add traces for each metric
for i, scenario in enumerate(df['scenario'].unique()):
    for j, metric in enumerate(metrics):
        scenario_data = df[df['scenario'] == scenario].sort_values(by=metric)
        
        # Add trace with the coloured bars based on the color_mapping
        colors = [color_mapping[x] for x in scenario_data['x_axis']]
        fig.add_trace(
            go.Bar(x=scenario_data['x_axis'], y=scenario_data[metric], name=f"{metric} Scenario {scenario}", marker_color=colors),
            row=i+1, col=j+1
        )

# Update layout
fig.update_layout(height=800*num_scenarios, width=1200, title_text="Metric Plots by Scenario", showlegend=False)

# Change the background colour of the plot and make as a grid
fig.update_layout(plot_bgcolor='white')

# Show plot
fig.show()
```
