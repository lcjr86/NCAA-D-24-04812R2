```python
import pandas as pd

import plotly.graph_objects as go
import plotly.express as px
#make_subplots
from plotly.subplots import make_subplots
```

```python
data_name = 'ETHUSDT_1h'
# from_date = '2021-01-01'
# until_date = '2022-12-31'
time_delay = 168
slice_size = 336
```

```python
df_features = pd.read_csv(f'../results/processed_log_return_wtmra_0_{data_name}_time_delay_{time_delay}_slice_size_{slice_size}/features/features_{data_name}_time_delay_{time_delay}_slice_size_{slice_size}.csv')
```

```python
# plot the features with plotly in multiples subplots
# data column : end_date
# Features columns : 'connected_components_entropy', 'loops_entropy', 'voids_entropy', 'connected_components_amplitude', 'loops_amplitude', 'voids_amplitude', 'connected_components_number_of_points', 'loops_number_of_points', 'voids_number_of_points'

initial_slice_position = 90

# Highlight a specific index range
highlight_start_idx = initial_slice_position
highlight_end_idx = initial_slice_position

# Convert the highlight_start_idx and highlight_end_idx to the corresponding date
highlight_start_date = df_features['end_date'][highlight_start_idx]
highlight_end_date = df_features['end_date'][highlight_end_idx]

fig = make_subplots(rows=9, cols=1, subplot_titles=('Entropy: Connected components [0]', 'Entropy: Loops [1]', 'Entropy: Voids [2]', 'Amplitude: Connected components [0]', 'Amplitude: Loops [1]', 'Amplitude: Voids [2]', 'Number of points: Connected components [0]', 'Number of points: Loops [1]', 'Number of points: Voids [2]'))

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['connected_components_entropy'], name='Entropy: Connected components [0]'), row=1, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['connected_components_entropy'].min(), x1=highlight_start_date, y1=df_features['connected_components_entropy'].max(),line=dict(color="Red",width=2),row=1, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['loops_entropy'], name='Entropy: Loops [1]'), row=2, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['loops_entropy'].min(), x1=highlight_start_date, y1=df_features['loops_entropy'].max(),line=dict(color="Red",width=2),row=2, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['voids_entropy'], name='Entropy: Voids [2]'), row=3, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['voids_entropy'].min(), x1=highlight_start_date, y1=df_features['voids_entropy'].max(),line=dict(color="Red",width=2),row=3, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['connected_components_amplitude'], name='Amplitude: Connected components [0]'), row=4, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['connected_components_amplitude'].min(), x1=highlight_start_date, y1=df_features['connected_components_amplitude'].max(),line=dict(color="Red",width=2),row=4, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['loops_amplitude'], name='Amplitude: Loops [1]'), row=5, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['loops_amplitude'].min(), x1=highlight_start_date, y1=df_features['loops_amplitude'].max(),line=dict(color="Red",width=2),row=5, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['voids_amplitude'], name='Amplitude: Voids [2]'), row=6, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['voids_amplitude'].min(), x1=highlight_start_date, y1=df_features['voids_amplitude'].max(),line=dict(color="Red",width=2),row=6, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['connected_components_number_of_points'], name='Number of points: Connected components [0]'), row=7, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['connected_components_number_of_points'].min(), x1=highlight_start_date, y1=df_features['connected_components_number_of_points'].max(),line=dict(color="Red",width=2),row=7, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['loops_number_of_points'], name='Number of points: Loops [1]'), row=8, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['loops_number_of_points'].min(), x1=highlight_start_date, y1=df_features['loops_number_of_points'].max(),line=dict(color="Red",width=2),row=8, col=1)

fig.add_trace(go.Scatter(x=df_features['end_date'], y=df_features['voids_number_of_points'], name='Number of points: Voids [2]'), row=9, col=1)
fig.add_shape(type="line", x0=highlight_start_date, y0=df_features['voids_number_of_points'].min(), x1=highlight_start_date, y1=df_features['voids_number_of_points'].max(),line=dict(color="Red",width=2),row=9, col=1)

# Remove legend
fig.update_layout(showlegend=False)

fig.update_layout(height=1500, width=1000, title_text="Features")
fig.show()
```
